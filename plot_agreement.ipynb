```python
import pandas as pd
import plotly.express as px

pd.options.plotting.backend = "plotly"


df = pd.read_csv("data/language_agreement_tokens_para_text.csv")
total_len = df.num_paragraphs.sum()

data = []

for size in df["size"]:
    if size > 110:
        break
    sub_df = df[df["size"] > size]
    data.append(
        {
            "size_threshold": size,
            "total_text_tokens": (sub_df.num_paragraphs * sub_df["size"]).sum(),
        }
    )


threshold_df = pd.DataFrame(data)
threshold_df.plot(x="size_threshold", y="total_text_tokens")
```
