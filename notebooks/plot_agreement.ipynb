# Plott forhold mellom størrelses-terksel for filtrering og datamengde beholdt 

## For antall tegn (str.len)
Kutter alt hvor avsnitt-lengden er under terskelen

```python
import pandas as pd

pd.options.plotting.backend = "plotly"


df = pd.read_csv("../data/language_agreement_len_para_text.csv")
total_len = df.num_paragraphs.sum()

data = []

for size in df["size"]:
    if size > 110:
        break
    sub_df = df[df["size"] > size]
    data.append(
        {
            "size_threshold": size,
            "total_text_len": (sub_df.num_paragraphs * sub_df["size"]).sum(),
        }
    )


threshold_df = pd.DataFrame(data)
threshold_df.plot(x="size_threshold", y="total_text_len")
```

## For antall tokens

Kutter alt hvor avsnittlengde er under terskelen

```python
df = pd.read_csv("../data/language_agreement_tokens_para_text.csv")
total_len = df.num_paragraphs.sum()

data = []

for size in df["size"]:
    if size > 110:
        break
    sub_df = df[df["size"] > size]
    data.append(
        {
            "size_threshold": size,
            "total_text_tokens": (sub_df.num_paragraphs * sub_df["size"]).sum(),
        }
    )


threshold_df = pd.DataFrame(data)
threshold_df.plot(x="size_threshold", y="total_text_tokens")
```

Kutter alt hvor avsnittlengde er over terskelen

```python
df = pd.read_csv("../data/language_agreement_tokens_para_text.csv")
total_len = df.num_paragraphs.sum()

data = []

for size in df["size"].sort_values(ascending=False):
    if size < 110:
        break
    sub_df = df[df["size"] < size]
    data.append(
        {
            "size_threshold": size,
            "total_text_tokens": (sub_df.num_paragraphs * sub_df["size"]).sum(),
        }
    )


threshold_df = pd.DataFrame(data)
threshold_df.plot(x="size_threshold", y="total_text_tokens")
```

```python
threshold_df = threshold_df[threshold_df.size_threshold < 10_000]
threshold_df.plot(x="size_threshold", y="total_text_tokens")
```

# Plott forhold mellom enighet mellom LID-modeller og lengde på avsnitt

## For antall tegn (str.len)

```python
df = pd.read_csv("../data/language_agreement_len_para_text.csv")
df["agreement_pct"] = df["agreeing_paragraphs"] / df["num_paragraphs"]
print(len(df))

df = df[df["num_paragraphs"] > 3]
print(len(df))

df.plot(x="size", y="agreement_pct")
```
